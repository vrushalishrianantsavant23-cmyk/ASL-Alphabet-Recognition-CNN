# Lab 1.1: Perceptron from Scratch

## Objective: Implement a single-layer perceptron using NumPy to classify AND/OR gates.

```python
import numpy as np

# Input data for AND gate
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0, 0, 0, 1])

# Initialize weights
w = np.zeros(X.shape[1])
bias = 0
lr = 0.1

def activation(x):
    return 1 if x >= 0 else 0

# Perceptron Learning
for epoch in range(10):
    for i in range(len(X)):
        z = np.dot(X[i], w) + bias
        y_hat = activation(z)
        error = y[i] - y_hat
        w += lr * error * X[i]
        bias += lr * error
print("Trained weights:", w)
print("Trained bias:", bias)
```
